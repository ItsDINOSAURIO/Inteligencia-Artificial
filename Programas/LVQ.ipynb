```python
# Importamos las librerías necesarias
import math
import matplotlib.pyplot as plt
 
# Inicialización de los pesos de las neuronas (vectores prototipo)
# Será necsario revisar a detalle lo expresado en el Capítulo 16
# Neurona 1 (Navegación)
w1 = [5.5, 16, 2, 3.8]
# Neurona 2 (Interacción Social)
w2 = [11, 55, 3, 7.2]
# Neurona 3 (Compras)
w3 = [22, 29, 2, 13.5]
# Neurona 4 (Consumo de Contenidos)
w4 = [17, 6, 1, 22.0]
```

```python
# Tasa de aprendizaje
alpha = 0.2
 
# Lista para guardar la evolución de los pesos
w1_history = [w1.copy()]
w2_history = [w2.copy()]
w3_history = [w3.copy()]
w4_history = [w4.copy()]
```

```python
# Conjunto de datos de entrenamiento (lista de ejemplos)
# Cada ejemplo es una tupla con el vector de entrada y su clase correcta
# Clase 1: Navegación, Clase 2: Interacción Social, Clase 3: Compras, Clase 4: Consumo de Contenidos
training_data = [
    ([23, 30, 2, 14], 3),  # Compras
    ([12, 50, 3, 8], 2),   # Interacción Social
    ([6, 15, 2, 4], 1),    # Navegación
    ([18, 5, 1, 20], 4),   # Consumo de Contenidos
    ([24, 28, 2, 15], 3),  # Compras
    ([10, 52, 3, 9], 2),   # Interacción Social
    ([5, 14, 2, 3.5], 1),  # Navegación
    ([16, 7, 1, 21], 4)    # Consumo de Contenidos
]
```

```python
# Número de épocas (iteraciones sobre el conjunto de datos)
epochs = 5
 
# Ciclo de entrenamiento
for epoch in range(epochs):
    print(f"\n--- Época {epoch + 1} ---")
    for x, correct_class in training_data:
        # Cálculo de las distancias euclidianas entre el vector de entrada y cada neurona
        d1 = math.sqrt(sum((x[i] - w1[i])**2 for i in range(4)))
        d2 = math.sqrt(sum((x[i] - w2[i])**2 for i in range(4)))
        d3 = math.sqrt(sum((x[i] - w3[i])**2 for i in range(4)))
        d4 = math.sqrt(sum((x[i] - w4[i])**2 for i in range(4)))
 
        # Determinamos la neurona ganadora
        distances = [d1, d2, d3, d4]
        min_distance = min(distances)
        winning_neuron = distances.index(min_distance) + 1  # Sumamos 1 porque los índices empiezan en 0
 
        # Actualizamos los pesos de acuerdo al resultado
        if winning_neuron == correct_class:
            # Acercamos los pesos de la neurona ganadora al vector de entrada
            print(f"\nEjemplo {x} clasificado correctamente por la Neurona {winning_neuron}.")
            if winning_neuron == 1:
                for i in range(4):
                    w1[i] = w1[i] + alpha * (x[i] - w1[i])
            elif winning_neuron == 2:
                for i in range(4):
                    w2[i] = w2[i] + alpha * (x[i] - w2[i])
            elif winning_neuron == 3:
                for i in range(4):
                    w3[i] = w3[i] + alpha * (x[i] - w3[i])
            elif winning_neuron == 4:
                for i in range(4):
                    w4[i] = w4[i] + alpha * (x[i] - w4[i])
        else:
            # Alejamos los pesos de la neurona ganadora del vector de entrada
            print(f"\nEjemplo {x} clasificado incorrectamente por la Neurona {winning_neuron}.")
            if winning_neuron == 1:
                for i in range(4):
                    w1[i] = w1[i] - alpha * (x[i] - w1[i])
            elif winning_neuron == 2:
                for i in range(4):
                    w2[i] = w2[i] - alpha * (x[i] - w2[i])
            elif winning_neuron == 3:
                for i in range(4):
                    w3[i] = w3[i] - alpha * (x[i] - w3[i])
            elif winning_neuron == 4:
                for i in range(4):
                    w4[i] = w4[i] - alpha * (x[i] - w4[i])
```

```python
#Actualizada
# Número de épocas (iteraciones sobre el conjunto de datos)
epochs = 5
 
# Ciclo de entrenamiento
for epoch in range(epochs):
    print(f"\n--- Época {epoch + 1} ---")
    for x, correct_class in training_data:
        # Cálculo de las distancias euclidianas entre el vector de entrada y cada neurona
        d1 = math.sqrt(sum((x[i] - w1[i])**2 for i in range(4)))
        d2 = math.sqrt(sum((x[i] - w2[i])**2 for i in range(4)))
        d3 = math.sqrt(sum((x[i] - w3[i])**2 for i in range(4)))
        d4 = math.sqrt(sum((x[i] - w4[i])**2 for i in range(4)))
 
        # Determinamos la neurona ganadora
        distances = [d1, d2, d3, d4]
        min_distance = min(distances)
        winning_neuron = distances.index(min_distance) + 1  # Sumamos 1 porque los índices empiezan en 0
 
        # Actualizamos los pesos de acuerdo al resultado
        if winning_neuron == correct_class:
            # Acercamos los pesos de la neurona ganadora al vector de entrada
            print(f"\nEjemplo {x} clasificado correctamente por la Neurona {winning_neuron}.")
            if winning_neuron == 1:
                for i in range(4):
                    w1[i] = w1[i] + alpha * (x[i] - w1[i])
            elif winning_neuron == 2:
                for i in range(4):
                    w2[i] = w2[i] + alpha * (x[i] - w2[i])
            elif winning_neuron == 3:
                for i in range(4):
                    w3[i] = w3[i] + alpha * (x[i] - w3[i])
            elif winning_neuron == 4:
                for i in range(4):
                    w4[i] = w4[i] + alpha * (x[i] - w4[i])
        else:
            # Alejamos los pesos de la neurona ganadora del vector de entrada
            print(f"\nEjemplo {x} clasificado incorrectamente por la Neurona {winning_neuron}.")
            if winning_neuron == 1:
                for i in range(4):
                    w1[i] = w1[i] - alpha * (x[i] - w1[i])
            elif winning_neuron == 2:
                for i in range(4):
                    w2[i] = w2[i] - alpha * (x[i] - w2[i])
            elif winning_neuron == 3:
                for i in range(4):
                    w3[i] = w3[i] - alpha * (x[i] - w3[i])
            elif winning_neuron == 4:
                for i in range(4):
                    w4[i] = w4[i] - alpha * (x[i] - w4[i])
 
        # Guardamos los pesos actuales en el historial
        w1_history.append(w1.copy())
        w2_history.append(w2.copy())
        w3_history.append(w3.copy())
        w4_history.append(w4.copy())
```

```python
# Después del entrenamiento, podemos visualizar cómo han cambiado los pesos
# Para simplificar, vamos a graficar solo el primer peso (Tiempo de Sesión) de cada neurona
# Cmo se indica en el libro "Neurla Network"
 
# Creamos una lista con los números de iteración
iterations = list(range(len(w1_history)))
 
# Extraemos el primer peso de cada neurona en cada iteración
w1_values = [w[0] for w in w1_history]
w2_values = [w[0] for w in w2_history]
w3_values = [w[0] for w in w3_history]
w4_values = [w[0] for w in w4_history]
 
# Graficamos la evolución del primer peso de cada neurona
plt.figure(figsize=(10, 6))
plt.plot(iterations, w1_values, label='Neurona 1 (Navegación)')
plt.plot(iterations, w2_values, label='Neurona 2 (Interacción Social)')
plt.plot(iterations, w3_values, label='Neurona 3 (Compras)')
plt.plot(iterations, w4_values, label='Neurona 4 (Consumo de Contenidos)')
plt.xlabel('Iteraciones')
plt.ylabel('Valor del Peso (Tiempo de Sesión)')
plt.title('Evolución del Peso del Tiempo de Sesión en Cada Neurona')
plt.legend()
plt.grid(True)
plt.show()
```
